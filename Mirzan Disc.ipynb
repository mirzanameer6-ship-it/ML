```python
from sklearn.tree import DecisionTreeClassifier
```

```python
x=[
    [20000 , 550],
    [25000 , 600],
    [30000 , 650],
    [35000 , 700],
    [40000 , 750],
    [45000 , 800],
    [50000 , 850]
]

y=[0 , 0 , 0 , 1 , 1 ,1 , 1]

model = DecisionTreeClassifier()
model.fit(x,y)
new_customer= [[40000 , 750]]

prediction = model.predict(new_customer)

if prediction[0]==1:
    print("Loan Approved")
else:
    print("Loan Rejected")
```
